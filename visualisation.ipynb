# VISUALISATION 


```python
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
random = pd.read_csv("logs/eval/random/default/random_eval_default.csv")
```

```python
def generate_elbow_plot(df):

    plt.plot(df["episode"], df["score"], 'o-', color = 'orange')
    plt.xlabel("episode")
    plt.ylabel("score")
    plt.title("score visualisation")
    plt.show()
```

```python
generate_elbow_plot(random)
```

```python
heuristic = pd.read_csv("logs/eval/heuristic/default/heuristic_eval_default.csv")
generate_elbow_plot(heuristic)
```
